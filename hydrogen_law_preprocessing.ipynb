```python
import re
import json
import os
from pathlib import Path

import pandas as pd
from pypdf import PdfReader
```

```python
from pypdf import PdfReader

def find_body_start_page(pdf_path):
    reader = PdfReader(pdf_path)

    for page_index, page in enumerate(reader.pages):
        text = page.extract_text() or ""

        # 실제 본문은 제1조(목적) 뒤에 "이 법은", "이 영은", "이 규칙은"이 이어짐
        if re.search(
            r"제\s*1\s*조\s*\(\s*목적\s*\)\s*(이\s*법은|이\s*영은|이\s*규칙은)",
            text
        ):
            return page_index

    return None
```

```python
print(type(pdf_files))
print(type(pdf_files[0]))
print(pdf_files[0])
```

```python
body_start_results = []

for pdf_path in pdf_files:

    start_index = find_body_start_page(pdf_path)

    body_start_results.append({
        "file_name": pdf_path,
        "start_index": start_index,
        "start_page": start_index + 1 if start_index is not None else None
    })

for result in body_start_results:
    print(
        result["file_name"],
        "→ 본문 시작 PDF 페이지:",
        result["start_page"]
    )
```

```python
all_body_texts = {}

for result in body_start_results:
    file_name = result["file_name"]
    start_index = result["start_index"]

    if start_index is None:
        print("본문 시작 페이지를 못 찾음:", file_name)
        continue

    reader = PdfReader(f"/content/{file_name}")

    body_text = ""

    for page in reader.pages[start_index:]:
        text = page.extract_text() or ""
        body_text += "\n" + text

    all_body_texts[file_name] = body_text

    print(
        file_name,
        "→ 본문 글자 수:",
        len(body_text)
    )
```

```python
import re

def clean_law_text(text):
    cleaned_lines = []

    for line in text.splitlines():
        line = line.strip()

        if not line:
            continue

        # 반복 머리말 제거
        if "법제처" in line and "국가법령정보센터" in line:
            continue

        # 페이지마다 반복되는 법령명 제거
        if line in [
            "수소경제 육성 및 수소 안전관리에 관한 법률",
            "수소경제 육성 및 수소 안전관리에 관한 법률 시행령",
            "수소경제 육성 및 수소 안전관리에 관한 법률 시행규칙",
            "고압가스 안전관리법",
            "고압가스 안전관리법 시행령",
            "고압가스 안전관리법 시행규칙",
            "액화석유가스의 안전관리 및 사업법",
            "액화석유가스의 안전관리 및 사업법 시행령",
            "액화석유가스의 안전관리 및 사업법 시행규칙",
        ]:
            continue

        cleaned_lines.append(line)

    cleaned_text = "\n".join(cleaned_lines)

    # 여러 칸 공백 정리
    cleaned_text = re.sub(r"[ \t]+", " ", cleaned_text)

    # 빈 줄 과다 반복 정리
    cleaned_text = re.sub(r"\n{3,}", "\n\n", cleaned_text)

    return cleaned_text.strip()


cleaned_body_texts = {}

for file_name, body_text in all_body_texts.items():
    cleaned_text = clean_law_text(body_text)
    cleaned_body_texts[file_name] = cleaned_text

    print(
        file_name,
        "→ 정리 전:", len(body_text),
        "/ 정리 후:", len(cleaned_text)
    )
```

```python
ARTICLE_PATTERN = re.compile(
    r"(?m)^제\s*(\d+)\s*조(?:의\s*(\d+))?\s*\(([^)\n]+)\)"
)

for file_name, text in cleaned_body_texts.items():
    matches = list(ARTICLE_PATTERN.finditer(text))

    print("\n" + "=" * 100)
    print(file_name)
    print("찾은 조문 수:", len(matches))
    print("앞 10개:", [m.group(0) for m in matches[:10]])
    print("마지막 5개:", [m.group(0) for m in matches[-5:]])
```

```python
main_body_texts = {}

for file_name, text in cleaned_body_texts.items():
    # 줄 시작이 '부칙'인 첫 지점부터 모두 제외
    appendix_match = re.search(
        r"(?m)^\s*부\s*칙(?:\s*<[^>\n]*>)?.*$",
        text
    )

    if appendix_match:
        main_text = text[:appendix_match.start()].strip()
    else:
        main_text = text.strip()

    main_body_texts[file_name] = main_text

    matches = list(ARTICLE_PATTERN.finditer(main_text))

    print(
        file_name,
        "→ 본문 조문 수:",
        len(matches),
        "/ 마지막 조문:",
        matches[-1].group(0) if matches else "없음"
    )
```

```python
all_articles = []

for file_name, text in main_body_texts.items():
    matches = list(ARTICLE_PATTERN.finditer(text))

    for i, match in enumerate(matches):
        start = match.start()

        if i + 1 < len(matches):
            end = matches[i + 1].start()
        else:
            end = len(text)

        article_number = match.group(1)
        sub_article_number = match.group(2)
        title = match.group(3).strip()
        content = text[start:end].strip()

        article_label = f"제{article_number}조"
        if sub_article_number:
            article_label += f"의{sub_article_number}"

        all_articles.append({
            "file_name": file_name,
            "article": article_label,
            "title": title,
            "content": content
        })

print("전체 조문 수:", len(all_articles))

for article in all_articles[:5]:
    print("\n" + "=" * 80)
    print(article["file_name"])
    print(article["article"], article["title"])
    print(article["content"][:500])
```

```python
import re

CHAPTER_PATTERN = re.compile(
    r"(?m)^\s*제\s*\d+\s*장(?:의\s*\d+)?\s+.*$"
)

for article in all_articles:
    article["content"] = CHAPTER_PATTERN.sub(
        "",
        article["content"]
    )

    article["content"] = re.sub(
        r"\n{3,}",
        "\n\n",
        article["content"]
    ).strip()

print("장 제목 제거 완료")

for article in all_articles[:5]:
    print("\n" + "=" * 80)
    print(article["article"], article["title"])
    print(article["content"][-300:])
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="BAAI/bge-m3",
    model_kwargs={
        "device": "cuda"
    },
    encode_kwargs={
        "normalize_embeddings": True,
        "batch_size": 16
    }
)

print("BGE-M3 로드 완료!")
```

```python
vector = embedding_model.embed_query(
    "수소용품 제조사업 허가 기준은 무엇인가요?"
)

print("벡터 길이:", len(vector))
print("앞의 값 5개:", vector[:5])
```

```python
from langchain_core.documents import Document

docs = []

for article in all_articles:
    searchable_text = (
        f"법령 파일: {article['file_name']}\n"
        f"조문: {article['article']}\n"
        f"제목: {article['title']}\n"
        f"내용:\n{article['content']}"
    )

    docs.append(
        Document(
            page_content=searchable_text,
            metadata={
                "file_name": article["file_name"],
                "article": article["article"],
                "title": article["title"]
            }
        )
    )

print("Document 수:", len(docs))
```

```python
!pip install -q langchain-chroma chromadb
```

```python
from langchain_chroma import Chroma

vectordb = Chroma.from_documents(
    documents=docs,
    embedding=embedding_model,
    collection_name="korean_hydrogen_laws",
    persist_directory="/content/hydrogen_chroma_db"
)

print("ChromaDB 생성 완료!")
print("저장된 문서 수:", vectordb._collection.count())
```

```python
query = "수소용품 제조사업 허가는 어떻게 받아야 하나요?"

results = vectordb.similarity_search(
    query,
    k=5
)

for i, doc in enumerate(results, start=1):
    print("\n" + "=" * 80)
    print("검색 결과", i)
    print("파일:", doc.metadata["file_name"])
    print("조문:", doc.metadata["article"])
    print("제목:", doc.metadata["title"])
    print(doc.page_content[:1000])
```
